```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import numpy as np

# Input features এবং target আলাদা করা
X = df.drop(columns=["Diabetes_binary", "ID"], errors="ignore")
y = df["Diabetes_binary"]

# প্রথমে 80% training, 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Training data থেকে validation data আলাদা করা
X_train_dl, X_val_dl, y_train_dl, y_val_dl = train_test_split(
    X_train, y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

# Feature scaling
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_dl)
X_val_scaled = scaler.transform(X_val_dl)
X_test_scaled = scaler.transform(X_test)

# Target-কে NumPy array-তে রূপান্তর
y_train_dl = np.asarray(y_train_dl)
y_val_dl = np.asarray(y_val_dl)
y_test_dl = np.asarray(y_test)

print("Training:", X_train_scaled.shape)
print("Validation:", X_val_scaled.shape)
print("Testing:", X_test_scaled.shape)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout, Input

# একই ধরনের ফল পুনরায় পাওয়ার সুবিধার জন্য
tf.keras.utils.set_random_seed(42)

# Deep Learning model
caretrust_dl = Sequential([
    Input(shape=(X_train_scaled.shape[1],)),

    Dense(64, activation="relu"),
    Dropout(0.2),

    Dense(32, activation="relu"),
    Dropout(0.2),

    Dense(1, activation="sigmoid")
])

# Model compile করা
caretrust_dl.compile(
    optimizer="adam",
    loss="binary_crossentropy",
    metrics=["accuracy"]
)

# Model-এর গঠন দেখা
caretrust_dl.summary()
```

```python
early_stop = tf.keras.callbacks.EarlyStopping(
    monitor="val_loss",
    patience=3,
    restore_best_weights=True
)

history = caretrust_dl.fit(
    X_train_scaled,
    y_train_dl,
    validation_data=(X_val_scaled, y_val_dl),
    epochs=20,
    batch_size=512,
    callbacks=[early_stop],
    verbose=1
)
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    roc_auc_score
)

# Test data দিয়ে probability বের করা
y_prob_dl = caretrust_dl.predict(X_test_scaled).ravel()

# Probability থেকে class নির্ধারণ
y_pred_dl = (y_prob_dl >= 0.5).astype(int)

print("Deep Learning Accuracy:",
      round(accuracy_score(y_test_dl, y_pred_dl), 4))

print("\nClassification Report:")
print(classification_report(
    y_test_dl, y_pred_dl, zero_division=0
))

print("ROC-AUC:",
      round(roc_auc_score(y_test_dl, y_prob_dl), 4))
```

```python
# D. CareTrust - Step 3: Probability Calibration

import numpy as np
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import brier_score_loss, roc_auc_score

# Check required variables
required = [
    "caretrust_dl", "X_val_scaled", "y_val_dl",
    "val_prob_dl", "X_test_scaled", "y_test_dl"
]

missing = [name for name in required if name not in globals()]
if missing:
    raise NameError(f"Missing variables: {missing}")

# 1. Get validation probabilities
eps = 1e-7

val_p = np.asarray(val_prob_dl).reshape(-1)
val_p = np.clip(val_p, eps, 1 - eps)

# 2. Convert probabilities to logits
val_logits = np.log(val_p / (1 - val_p)).reshape(-1, 1)

# 3. Fit calibration model using validation data only
calibrator = LogisticRegression()
calibrator.fit(val_logits, np.asarray(y_val_dl).reshape(-1))

# 4. Predict on the untouched test set
test_prob_raw = caretrust_dl.predict(
    X_test_scaled, verbose=0
).reshape(-1)

test_prob_raw = np.clip(test_prob_raw, eps, 1 - eps)

# 5. Calibrate test probabilities
test_logits = np.log(
    test_prob_raw / (1 - test_prob_raw)
).reshape(-1, 1)

test_prob_cal = calibrator.predict_proba(
    test_logits
)[:, 1]

# 6. Compare raw and calibrated probabilities
y_true = np.asarray(y_test_dl).reshape(-1)

print("Raw Brier Score:",
      round(brier_score_loss(y_true, test_prob_raw), 5))

print("Calibrated Brier Score:",
      round(brier_score_loss(y_true, test_prob_cal), 5))

print("Raw ROC-AUC:",
      round(roc_auc_score(y_true, test_prob_raw), 5))

print("Calibrated ROC-AUC:",
      round(roc_auc_score(y_true, test_prob_cal), 5))

print("test_prob_raw exists:", "test_prob_raw" in globals())
print("test_prob_cal exists:", "test_prob_cal" in globals())
```

```python
# D. CareTrust - Step 4: Uncertainty Estimation

import numpy as np
import matplotlib.pyplot as plt

# Make sure Step 3 has been completed
if "test_prob_cal" not in globals():
    raise NameError(
        "Please run Step 3 (Probability Calibration) first."
    )

# 1. Calculate uncertainty from calibrated probabilities
test_prob_cal = np.asarray(test_prob_cal).reshape(-1)

uncertainty_test = 1 - np.abs(
    2 * test_prob_cal - 1
)

# 2. Keep uncertainty within [0, 1]
uncertainty_test = np.clip(
    uncertainty_test, 0, 1
)

# 3. Display summary
print("Number of test predictions:",
      len(uncertainty_test))

print("Mean uncertainty:",
      round(float(np.mean(uncertainty_test)), 4))

print("Minimum uncertainty:",
      round(float(np.min(uncertainty_test)), 4))

print("Maximum uncertainty:",
      round(float(np.max(uncertainty_test)), 4))

# 4. Show example predictions
print("\nFirst 10 predictions:")
for i in range(min(10, len(test_prob_cal))):
    print(
        f"Probability={test_prob_cal[i]:.4f}, "
        f"Uncertainty={uncertainty_test[i]:.4f}"
    )

# 5. Plot uncertainty distribution
plt.figure(figsize=(8, 5))
plt.hist(uncertainty_test, bins=20, edgecolor="black")
plt.xlabel("Uncertainty Score")
plt.ylabel("Number of Predictions")
plt.title("D. CareTrust: Uncertainty Distribution")
plt.tight_layout()
plt.show()
```

```python
# D. CareTrust - Step 5: Reliability Score & Selective Review

import numpy as np
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# 1. Check required variables
required = [
    "calibrator", "val_prob_dl", "y_val_dl",
    "test_prob_cal", "uncertainty_test", "y_test_dl"
]

missing = [v for v in required if v not in globals()]
if missing:
    raise NameError(
        f"Missing variables: {missing}. Run previous steps first."
    )

# 2. Calibrate validation probabilities
eps = 1e-7
val_p = np.clip(
    np.asarray(val_prob_dl).reshape(-1),
    eps, 1 - eps
)

val_logits = np.log(val_p / (1 - val_p)).reshape(-1, 1)
val_prob_cal = calibrator.predict_proba(val_logits)[:, 1]

# 3. Calculate validation uncertainty
val_uncertainty = 1 - np.abs(2 * val_prob_cal - 1)

# 4. Define a preliminary reliability score
val_trust_score = 1 - val_uncertainty

test_p = np.asarray(test_prob_cal).reshape(-1)
test_uncertainty = np.asarray(uncertainty_test).reshape(-1)
test_trust_score = 1 - test_uncertainty

# 5. Select threshold using validation data only
target_coverage = 0.80
uncertainty_threshold = np.quantile(
    val_uncertainty, target_coverage
)

# Lower uncertainty = accept; higher uncertainty = review
accepted = test_uncertainty <= uncertainty_threshold
reviewed = ~accepted

# 6. Generate predictions
y_true = np.asarray(y_test_dl).reshape(-1).astype(int)
y_pred = (test_p >= 0.5).astype(int)

# 7. Evaluate all test predictions
print("----- ALL TEST PREDICTIONS -----")
print("Overall accuracy:",
      round(accuracy_score(y_true, y_pred), 4))

# 8. Evaluate accepted predictions
print("\n----- ACCEPTED PREDICTIONS -----")
print("Accepted count:", int(accepted.sum()))
print("Review count:", int(reviewed.sum()))
print("Actual coverage:",
      round(float(accepted.mean()), 4))

if accepted.sum() > 0:
    print("Accepted accuracy:",
          round(accuracy_score(y_true[accepted], y_pred[accepted]), 4))
    print("Selective risk (error rate):",
          round(1 - accuracy_score(
              y_true[accepted], y_pred[accepted]
          ), 4))
    print("Accepted positive-class F1:",
          round(f1_score(
              y_true[accepted], y_pred[accepted],
              zero_division=0
          ), 4))
    print("Accepted positive-class recall:",
          round(recall_score(
              y_true[accepted], y_pred[accepted],
              zero_division=0
          ), 4))

print("\n----- REVIEW RULE -----")
print("Target coverage:", target_coverage)
print("Validation uncertainty threshold:",
      round(float(uncertainty_threshold), 4))
print("Mean test trust score:",
      round(float(test_trust_score.mean()), 4))
```

```python
# D. CareTrust - Step 8: Validation-based Threshold Optimization

import numpy as np
import pandas as pd
from sklearn.metrics import (
    f1_score, recall_score, precision_score,
    accuracy_score
)

# Check required variables
required = [
    "val_prob_dl", "y_val_dl", "calibrator",
    "test_prob_cal", "y_test_dl"
]
missing = [v for v in required if v not in globals()]
if missing:
    raise NameError(f"Missing variables: {missing}")

# 1. Calibrate validation probabilities
eps = 1e-7
val_p = np.clip(
    np.asarray(val_prob_dl).reshape(-1), eps, 1-eps
)
val_logits = np.log(val_p / (1-val_p)).reshape(-1, 1)
val_p_cal = calibrator.predict_proba(val_logits)[:, 1]

y_val = np.asarray(y_val_dl).reshape(-1).astype(int)

# 2. Calculate validation uncertainty
val_uncertainty = 1 - np.abs(2 * val_p_cal - 1)

# 3. Search thresholds using validation data only
min_coverage = 0.70
candidate_thresholds = np.unique(
    np.quantile(val_uncertainty, np.linspace(0, 1, 101))
)

candidates = []

for threshold in candidate_thresholds:
    accepted_val = val_uncertainty <= threshold
    coverage = accepted_val.mean()

    if coverage < min_coverage or accepted_val.sum() == 0:
        continue

    pred_val = (val_p_cal >= 0.5).astype(int)

    f1 = f1_score(
        y_val[accepted_val],
        pred_val[accepted_val],
        zero_division=0
    )

    recall = recall_score(
        y_val[accepted_val],
        pred_val[accepted_val],
        zero_division=0
    )

    candidates.append({
        "threshold": threshold,
        "coverage": coverage,
        "class1_f1": f1,
        "class1_recall": recall
    })

if not candidates:
    raise ValueError("No threshold met the minimum coverage.")

# Choose the threshold with the highest validation Class 1 F1
best = max(
    candidates,
    key=lambda row: (row["class1_f1"], row["coverage"])
)

best_threshold = best["threshold"]

print("Best validation threshold:",
      round(float(best_threshold), 4))
print("Validation coverage:",
      round(float(best["coverage"]), 4))
print("Validation Class 1 F1:",
      round(float(best["class1_f1"]), 4))
print("Validation Class 1 Recall:",
      round(float(best["class1_recall"]), 4))

# 4. Apply the chosen threshold to the test set
test_p = np.asarray(test_prob_cal).reshape(-1)
y_test = np.asarray(y_test_dl).reshape(-1).astype(int)

test_uncertainty = 1 - np.abs(2 * test_p - 1)
accepted_test = test_uncertainty <= best_threshold

test_pred = (test_p >= 0.5).astype(int)

print("\n--- FINAL TEST RESULTS ---")
print("Test coverage:",
      round(float(accepted_test.mean()), 4))
print("Accepted samples:", int(accepted_test.sum()))
print("Sent for review:", int((~accepted_test).sum()))

if accepted_test.sum() > 0:
    print("Accepted accuracy:",
          round(accuracy_score(
              y_test[accepted_test],
              test_pred[accepted_test]
          ), 4))
    print("Accepted Class 1 F1:",
          round(f1_score(
              y_test[accepted_test],
              test_pred[accepted_test],
              zero_division=0
          ), 4))
    print("Accepted Class 1 Recall:",
          round(recall_score(
              y_test[accepted_test],
              test_pred[accepted_test],
              zero_division=0
          ), 4))
    print("Accepted Class 1 Precision:",
          round(precision_score(
              y_test[accepted_test],
              test_pred[accepted_test],
              zero_division=0
          ), 4))

# Save selected threshold and mask for subsequent steps
caretrust_threshold = best_threshold
accepted = accepted_test
```

```python
rng = np.random.default_rng(42)
y = np.asarray(y_test_dl).astype(int)
n = len(y)
acc_mask = np.asarray(accepted).astype(bool)

pred_raw = (np.asarray(test_prob_raw) >= 0.5).astype(int)
pred_cal = (np.asarray(test_prob_cal) >= 0.5).astype(int)
pred_thr = (np.asarray(test_prob_cal) >= best_threshold).astype(int)

def prf(yy, pp):
    tp = np.sum((pp == 1) & (yy == 1)); fp = np.sum((pp == 1) & (yy == 0))
    fn = np.sum((pp == 0) & (yy == 1)); tn = np.sum((pp == 0) & (yy == 0))
    prec = tp / max(tp + fp, 1); rec = tp / max(tp + fn, 1)
    f1 = 2 * prec * rec / max(prec + rec, 1e-12)
    return {"Accuracy": (tp + tn) / max(len(yy), 1), "Precision": prec,
            "Recall": rec, "F1": f1}

configs = [
    ("A. DL Raw (t=0.5)",                      pred_raw, None),
    ("B. + Calibration (t=0.5)",               pred_cal, None),
    ("C. + Threshold optimization",            pred_thr, None),
    ("D. + Uncertainty rejection (CareTrust)", pred_thr, acc_mask),
]

B = 1000
boot = {name: {k: [] for k in ["Accuracy", "Precision", "Recall", "F1"]}
        for name, _, _ in configs}
for _ in range(B):
    idx = rng.integers(0, n, n)
    for name, pp, mk in configs:
        if mk is None:
            r = prf(y[idx], pp[idx])
        else:
            m = mk[idx]
            r = prf(y[idx][m], pp[idx][m])
        for k, v in r.items():
            boot[name][k].append(v)

out = []
for name, pp, mk in configs:
    full = prf(y, pp) if mk is None else prf(y[mk], pp[mk])
    row = {"Setup": name, "Coverage": round(1.0 if mk is None else mk.mean(), 3)}
    for k in ["Accuracy", "Precision", "Recall", "F1"]:
        lo, hi_ = np.percentile(boot[name][k], [2.5, 97.5])
        row[k] = f"{full[k]:.3f} [{lo:.3f}, {hi_:.3f}]"
    out.append(row)
ablation_df = pd.DataFrame(out)
print(ablation_df.to_string(index=False))

rej = ~acc_mask
print(f"\nRejected: {rej.sum()} cases ({rej.mean():.1%})")
print(f"Error rate  accepted: {(pred_thr[acc_mask] != y[acc_mask]).mean():.4f} | "
      f"rejected: {(pred_thr[rej] != y[rej]).mean():.4f}")
print(f"Diabetic share accepted: {y[acc_mask].mean():.3f} | rejected: {y[rej].mean():.3f}")

# DL vs LR: AUC পার্থক্যের paired bootstrap CI
Ba = 200
d_roc, d_pr = [], []
for _ in range(Ba):
    idx = rng.integers(0, n, n)
    yy = y[idx]
    d_roc.append(roc_auc_score(yy, np.asarray(test_prob_cal)[idx])
                 - roc_auc_score(yy, lr_plain_prob[idx]))
    d_pr.append(average_precision_score(yy, np.asarray(test_prob_cal)[idx])
                - average_precision_score(yy, lr_plain_prob[idx]))
for nm, d in [("ΔROC-AUC (DL − LR)", d_roc), ("ΔPR-AUC  (DL − LR)", d_pr)]:
    lo, hi_ = np.percentile(d, [2.5, 97.5])
    print(f"{nm}: {np.mean(d):+.4f}  95% CI [{lo:+.4f}, {hi_:+.4f}]")
```

```python
from scipy.stats import spearmanr
from sklearn.metrics import roc_auc_score

y = np.asarray(y_test_dl).astype(int)
p = np.asarray(test_prob_cal).reshape(-1)
pr = np.asarray(test_prob_raw).reshape(-1)
u = np.asarray(uncertainty_test).reshape(-1)
pred = (p >= best_threshold).astype(int)
P = y.sum()
neg = pred == 0
margin_unc = -np.abs(p - best_threshold)
rand_unc = np.random.default_rng(12345).random(len(y))

# ---- 1. u কি probability থেকে আলাদা কিছু জানায়? ----
print("Spearman(u, p_cal)       :", round(spearmanr(u, p)[0], 4))
print("Spearman(u, p_raw)       :", round(spearmanr(u, pr)[0], 4))
print("Spearman(u, 1-|2p-1|)    :", round(spearmanr(u, 1 - np.abs(2 * p - 1))[0], 4))
print("Missed-diabetic detection AUROC (শুধু predicted-negative কেসে):")
print("   u    :", round(roc_auc_score(y[neg], u[neg]), 4))
print("   p_cal:", round(roc_auc_score(y[neg], p[neg]), 4))

# ---- 2. Triage: referred কেস clinician দেখবেন (সঠিকভাবে, ধরে নেওয়া) ----
def triage(score, name, only_neg=False):
    rows = []
    pool = np.where(neg)[0] if only_neg else np.arange(len(y))
    for ref in [0.0, 0.10, 0.20, 0.30]:
        refer = np.zeros(len(y), bool)
        k = int(round(ref * len(y)))
        if k > 0:
            refer[pool[np.argsort(-score[pool])[:k]]] = True
        keep = ~refer
        rows.append({
            "Method": name, "Referral": ref,
            "Auto_error": (pred[keep] != y[keep]).mean(),
            "Silent_miss": np.sum(keep & (pred == 0) & (y == 1)) / P,
            "Ref_diab_share": y[refer].mean() if k > 0 else np.nan,
        })
    return pd.DataFrame(rows)

tri = pd.concat([
    triage(u, "CareTrust u (all)"),
    triage(u, "CareTrust u (pred-neg only)", only_neg=True),
    triage(margin_unc, "Baseline |p-thr|"),
    triage(rand_unc, "Random"),
], ignore_index=True)

for col, title in [("Silent_miss", "Silent miss rate (সব diabetic-এর কত অংশ মিস হলো)"),
                   ("Auto_error", "Auto error (স্বয়ংক্রিয় কেসে error rate)"),
                   ("Ref_diab_share", "Referred-এ diabetic-এর হার")]:
    print("\n" + title)
    print(tri.pivot(index="Referral", columns="Method",
                    values=col).round(3).to_string())
```

```python
from scipy.stats import spearmanr, rankdata
from sklearn.metrics import f1_score

d = np.load(ENS_PATH)
val_m, test_m = d["val"], d["test"]
yv = np.asarray(y_val_dl).astype(int)
yt = np.asarray(y_test_dl).astype(int)

pv, sdv = val_m.mean(0), val_m.std(0)
pt, sdt = test_m.mean(0), test_m.std(0)

# ---- 1. Ensemble মান ----
aucs = [roc_auc_score(yt, m) for m in test_m]
prs = [average_precision_score(yt, m) for m in test_m]
print(f"Members ROC-AUC: {np.mean(aucs):.4f} ± {np.std(aucs):.4f} | "
      f"PR-AUC: {np.mean(prs):.4f} ± {np.std(prs):.4f}")
print(f"Ensemble ROC-AUC {roc_auc_score(yt, pt):.4f} | "
      f"PR-AUC {average_precision_score(yt, pt):.4f} | "
      f"Brier {brier_score_loss(yt, pt):.5f} | ECE {ece_score(yt, pt):.4f}")

# ---- 2. Threshold validation set থেকে ----
ts = np.linspace(0.05, 0.60, 111)
thr = float(ts[int(np.argmax([f1_score(yv, (pv >= t).astype(int), zero_division=0)
                              for t in ts]))])
pred_v = (pv >= thr).astype(int)
pred_t = (pt >= thr).astype(int)
neg_t = pred_t == 0
print(f"Ensemble threshold (validation F1-max): {thr:.3f}")

# ---- 3. sd কি p-র বাইরে তথ্য দেয়? ----
print("\nSpearman(sd, p):", round(spearmanr(sdt, pt)[0], 4))
print("Missed-diabetic AUROC (predicted-negative কেসে)  "
      f"p: {roc_auc_score(yt[neg_t], pt[neg_t]):.4f} | "
      f"sd: {roc_auc_score(yt[neg_t], sdt[neg_t]):.4f}")

# ---- 4. Triage: p বনাম epistemic বনাম মিশ্র (weight validation থেকে) ----
def combo(p, s, w):
    return (1 - w) * rankdata(p) / len(p) + w * rankdata(s) / len(s)

def silent_miss(y, pred, score, ref):
    neg = np.where(pred == 0)[0]
    k = int(round(ref * len(y)))
    refer = np.zeros(len(y), bool)
    if k > 0:
        refer[neg[np.argsort(-score[neg])[:k]]] = True
    return np.sum(~refer & (pred == 0) & (y == 1)) / y.sum()

ws = [0, 0.25, 0.5, 0.75, 1.0]
val_scores = {w: silent_miss(yv, pred_v, combo(pv, sdv, w), 0.20) for w in ws}
w_star = min(val_scores, key=val_scores.get)
print("Validation silent-miss @20% referral by w:",
      {w: round(float(v), 3) for w, v in val_scores.items()}, "→ w* =", w_star)

rows = []
for name, w in [("p only (w=0)", 0.0), ("epistemic sd only (w=1)", 1.0),
                (f"combined (w*={w_star})", w_star)]:
    sc = combo(pt, sdt, w)
    rows.append({"Score": name,
                 **{f"ref {int(r * 100)}%": round(float(silent_miss(yt, pred_t, sc, r)), 3)
                    for r in [0.1, 0.2, 0.3]}})
print("\nTest silent-miss rate (no referral:",
      round(float(silent_miss(yt, pred_t, pt, 0.0)), 3), ")")
print(pd.DataFrame(rows).to_string(index=False))
```

```python
# ===== CareTrust: full pipeline (run in ONE Colab cell) =====

# 1. Setup: load saved deep-learning outputs and rebuild the data split
!pip install -q ucimlrepo lightgbm

import os, warnings, joblib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import drive
warnings.filterwarnings("ignore", message="X does not have valid feature names")

drive.mount("/content/drive")
DRIVE_DIR = "/content/drive/MyDrive/D_CareTrust_Backup/"
FIG_DIR = DRIVE_DIR + "figs/"
RES_DIR = DRIVE_DIR + "results/"
os.makedirs(FIG_DIR, exist_ok=True)
os.makedirs(RES_DIR, exist_ok=True)

saved = joblib.load(DRIVE_DIR + "important_variables.joblib")
globals().update(saved)

from ucimlrepo import fetch_ucirepo
from sklearn.model_selection import train_test_split

ds = fetch_ucirepo(id=891)
X_all = ds.data.features
y_all = ds.data.targets["Diabetes_binary"].values

SEED = 42
X_tv, X_te_raw, y_tv, y_te = train_test_split(
    X_all, y_all, test_size=0.2, random_state=SEED, stratify=y_all)
X_tr_raw, X_va_raw, y_tr, y_va = train_test_split(
    X_tv, y_tv, test_size=0.2, random_state=SEED, stratify=y_tv)

assert np.array_equal(y_te, y_test_dl) and np.array_equal(y_va, y_val_dl)
assert np.allclose(scaler.transform(X_tr_raw.values), X_train_scaled)
y_train_dl = y_tr
print("Split verified:", X_train_scaled.shape, X_val_scaled.shape, X_test_scaled.shape)

# 2. Metrics, plot style, LR baselines, comparison table
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             roc_auc_score, average_precision_score, brier_score_loss)

INK, MUTED, GRID = "#0b0b0b", "#52514e", "#e4e3df"
BLUE, ORANGE, AQUA, YELLOW = "#2a78d6", "#eb6834", "#1baf7a", "#eda100"
plt.rcParams.update({
    "font.size": 11, "text.color": INK, "axes.labelcolor": INK,
    "axes.edgecolor": MUTED, "xtick.color": MUTED, "ytick.color": MUTED,
    "axes.spines.top": False, "axes.spines.right": False,
    "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.8,
    "axes.axisbelow": True, "figure.facecolor": "white"})

def ece_score(y_true, p, n_bins=10):
    y_true = np.asarray(y_true).reshape(-1)
    p = np.asarray(p).reshape(-1)
    idx = np.digitize(p, np.linspace(0, 1, n_bins + 1)[1:-1])
    ece = 0.0
    for k in range(n_bins):
        m = idx == k
        if m.any():
            ece += m.mean() * abs(y_true[m].mean() - p[m].mean())
    return ece

def evaluate_model(name, y_true, probabilities, threshold=0.5):
    y_true = np.asarray(y_true).reshape(-1).astype(int)
    probabilities = np.asarray(probabilities).reshape(-1)
    pred = (probabilities >= threshold).astype(int)
    return {
        "Model": name, "Threshold": round(float(threshold), 3),
        "Accuracy": round(accuracy_score(y_true, pred), 4),
        "Precision_1": round(precision_score(y_true, pred, zero_division=0), 4),
        "Recall_1": round(recall_score(y_true, pred, zero_division=0), 4),
        "F1_1": round(f1_score(y_true, pred, zero_division=0), 4),
        "ROC_AUC": round(roc_auc_score(y_true, probabilities), 4),
        "PR_AUC": round(average_precision_score(y_true, probabilities), 4),
        "Brier": round(brier_score_loss(y_true, probabilities), 5),
        "ECE": round(ece_score(y_true, probabilities), 4)}

def prf(yy, pp):
    tp = np.sum((pp == 1) & (yy == 1)); fp = np.sum((pp == 1) & (yy == 0))
    fn = np.sum((pp == 0) & (yy == 1)); tn = np.sum((pp == 0) & (yy == 0))
    prec = tp / max(tp + fp, 1); rec = tp / max(tp + fn, 1)
    f1 = 2 * prec * rec / max(prec + rec, 1e-12)
    return {"Accuracy": (tp + tn) / max(len(yy), 1), "Precision": prec,
            "Recall": rec, "F1": f1}

def best_f1_threshold(y_true, p):
    ts = np.linspace(0.05, 0.95, 181)
    f1s = [f1_score(y_true, (p >= t).astype(int), zero_division=0) for t in ts]
    return float(ts[int(np.argmax(f1s))])

lr_bal = LogisticRegression(max_iter=1000, class_weight="balanced",
                            random_state=42).fit(X_train_scaled, y_train_dl)
lr_plain = LogisticRegression(max_iter=1000, random_state=42).fit(X_train_scaled, y_train_dl)
lr_bal_prob = lr_bal.predict_proba(X_test_scaled)[:, 1]
lr_plain_prob = lr_plain.predict_proba(X_test_scaled)[:, 1]
lr_bal_val = lr_bal.predict_proba(X_val_scaled)[:, 1]
lr_val_p = lr_plain.predict_proba(X_val_scaled)[:, 1]
lr_thr = best_f1_threshold(y_val_dl, lr_val_p)

rows = [
    evaluate_model("LR (balanced)", y_test_dl, lr_bal_prob),
    evaluate_model("LR (unweighted)", y_test_dl, lr_plain_prob),
    evaluate_model("LR (unweighted) + opt. threshold", y_test_dl, lr_plain_prob, lr_thr),
    evaluate_model("DL Raw", y_test_dl, test_prob_raw),
    evaluate_model("DL Calibrated", y_test_dl, test_prob_cal),
    evaluate_model("DL Calibrated + opt. threshold", y_test_dl, test_prob_cal,
                   threshold=best_threshold),
]
comparison_df = pd.DataFrame(rows)
print(comparison_df.to_string(index=False))

# 3. Calibration (reliability) curves
from sklearn.calibration import calibration_curve

curves = []
for name, p, col, ls in [
    ("DL Calibrated",   test_prob_cal, BLUE,   "-"),
    ("DL Raw",          test_prob_raw, ORANGE, "--"),
    ("LR (unweighted)", lr_plain_prob, AQUA,   "-"),
    ("LR (balanced)",   lr_bal_prob,   YELLOW, "--"),
]:
    frac_pos, mean_pred = calibration_curve(y_test_dl, p, n_bins=10, strategy="quantile")
    curves.append((name, mean_pred, frac_pos, col, ls))

hi = min(1.0, max(max(m.max(), f.max()) for _, m, f, _, _ in curves) + 0.05)
fig, ax = plt.subplots(figsize=(5.6, 5.6))
ax.plot([0, hi], [0, hi], ls=":", color=MUTED, lw=1.5, label="Perfect calibration")
for name, mean_pred, frac_pos, col, ls in curves:
    ax.plot(mean_pred, frac_pos, color=col, ls=ls, lw=2, marker="o", ms=6,
            mfc=col, mec="white", mew=1, label=name)
ax.set_xlim(0, hi); ax.set_ylim(0, hi)
ax.set_xlabel("Mean predicted probability")
ax.set_ylabel("Observed fraction of diabetic")
ax.set_title("Calibration curve (test set)", loc="left", fontsize=12)
ax.legend(frameon=False, loc="upper center", bbox_to_anchor=(0.5, -0.14), ncol=2)
fig.savefig(FIG_DIR + "calibration_curve.png", dpi=300, bbox_inches="tight")
plt.show()

# 4. Ablation (A-C) with bootstrap 95% CIs, DL vs LR significance
y_ab = np.asarray(y_test_dl).astype(int)
n_ab = len(y_ab)
pred_raw = (np.asarray(test_prob_raw) >= 0.5).astype(int)
pred_cal = (np.asarray(test_prob_cal) >= 0.5).astype(int)
pred_thr = (np.asarray(test_prob_cal) >= best_threshold).astype(int)
configs = [
    ("A. DL Raw (t=0.5)", pred_raw),
    ("B. + Calibration (t=0.5)", pred_cal),
    ("C. + Threshold optimization", pred_thr),
]
rng_ab = np.random.default_rng(42)
B = 1000
boot = {n: {k: [] for k in ["Accuracy", "Precision", "Recall", "F1"]} for n, _ in configs}
for _ in range(B):
    idx = rng_ab.integers(0, n_ab, n_ab)
    for name, pp in configs:
        for k, v in prf(y_ab[idx], pp[idx]).items():
            boot[name][k].append(v)

out = []
for name, pp in configs:
    full = prf(y_ab, pp)
    row = {"Setup": name}
    for k in ["Accuracy", "Precision", "Recall", "F1"]:
        lo, hi_ = np.percentile(boot[name][k], [2.5, 97.5])
        row[k] = f"{full[k]:.3f} [{lo:.3f}, {hi_:.3f}]"
    out.append(row)
ablation_df = pd.DataFrame(out)
print(ablation_df.to_string(index=False))

dl_p = np.asarray(test_prob_cal).reshape(-1)
d_roc, d_pr = [], []
for _ in range(200):
    idx = rng_ab.integers(0, n_ab, n_ab)
    yy = y_ab[idx]
    d_roc.append(roc_auc_score(yy, dl_p[idx]) - roc_auc_score(yy, lr_plain_prob[idx]))
    d_pr.append(average_precision_score(yy, dl_p[idx])
                - average_precision_score(yy, lr_plain_prob[idx]))
for nm, dd in [("Delta ROC-AUC (DL - LR)", d_roc), ("Delta PR-AUC  (DL - LR)", d_pr)]:
    lo, hi_ = np.percentile(dd, [2.5, 97.5])
    print(f"{nm}: {np.mean(dd):+.4f}  95% CI [{lo:+.4f}, {hi_:+.4f}]")

# 5. LightGBM baseline + paired bootstrap vs DL
import lightgbm as lgb

gbm = lgb.LGBMClassifier(
    n_estimators=3000, learning_rate=0.03, num_leaves=31, min_child_samples=50,
    subsample=0.8, subsample_freq=1, colsample_bytree=0.8, reg_lambda=1.0,
    random_state=42, verbose=-1)
gbm.fit(X_tr_raw, y_tr, eval_set=[(X_va_raw, y_va)], eval_metric="binary_logloss",
        callbacks=[lgb.early_stopping(100, verbose=False)])
gbm_prob = gbm.predict_proba(X_te_raw)[:, 1]
gbm_val = gbm.predict_proba(X_va_raw)[:, 1]
print("LightGBM trees used:", gbm.best_iteration_)

yt0 = np.asarray(y_test_dl).astype(int)
gbm_table = pd.DataFrame([{
    "Model": n, "ROC_AUC": round(roc_auc_score(yt0, p), 4),
    "PR_AUC": round(average_precision_score(yt0, p), 4),
    "Brier": round(brier_score_loss(yt0, p), 5), "ECE": round(ece_score(yt0, p), 4)}
    for n, p in [("LR (unweighted)", lr_plain_prob), ("LightGBM", gbm_prob),
                 ("DL Calibrated", dl_p)]])
print(gbm_table.to_string(index=False))

rng5 = np.random.default_rng(7)
d_roc, d_pr = [], []
for _ in range(200):
    idx = rng5.integers(0, len(yt0), len(yt0))
    yy = yt0[idx]
    d_roc.append(roc_auc_score(yy, dl_p[idx]) - roc_auc_score(yy, gbm_prob[idx]))
    d_pr.append(average_precision_score(yy, dl_p[idx])
                - average_precision_score(yy, gbm_prob[idx]))
for nm, dd in [("Delta ROC-AUC (DL - LightGBM)", d_roc), ("Delta PR-AUC  (DL - LightGBM)", d_pr)]:
    lo, hi_ = np.percentile(dd, [2.5, 97.5])
    print(f"{nm}: {np.mean(dd):+.4f}  95% CI [{lo:+.4f}, {hi_:+.4f}]")

# 6. Does the original uncertainty score add information beyond p?
from scipy.stats import spearmanr

if "uncertainty_test" in globals():
    u6 = np.asarray(uncertainty_test).reshape(-1)
    p6 = np.asarray(test_prob_cal).reshape(-1)
    y6 = np.asarray(y_test_dl).astype(int)
    neg6 = (p6 < best_threshold)
    print("Spearman(u, p_cal)     :", round(spearmanr(u6, p6)[0], 4))
    print("Spearman(u, 1-|2p-1|)  :", round(spearmanr(u6, 1 - np.abs(2 * p6 - 1))[0], 4))
    print("Missed-diabetic detection AUROC among predicted negatives:")
    print("   u     :", round(roc_auc_score(y6[neg6], u6[neg6]), 4))
    print("   p_cal :", round(roc_auc_score(y6[neg6], p6[neg6]), 4))
else:
    print("No saved uncertainty score found; skipping this diagnostic.")

# 7. Risk-band triage vs baselines
y_tri = np.asarray(y_test_dl).astype(int)
p_tri = np.asarray(test_prob_cal).reshape(-1)
pred_tri = (p_tri >= best_threshold).astype(int)
neg_idx = np.where(pred_tri == 0)[0]
P_tri = y_tri.sum()

def triage_metrics(score, ref, only_neg):
    pool = neg_idx if only_neg else np.arange(len(y_tri))
    k = int(round(ref * len(y_tri)))
    refer = np.zeros(len(y_tri), bool)
    if k > 0:
        refer[pool[np.argsort(-score[pool])[:k]]] = True
    keep = ~refer
    return {"Referral": ref,
            "Silent_miss": np.sum(keep & (pred_tri == 0) & (y_tri == 1)) / P_tri,
            "Auto_error": (pred_tri[keep] != y_tri[keep]).mean(),
            "Referred_diabetic_share": y_tri[refer].mean() if k > 0 else np.nan}

policies = {
    "Risk-band (p, predicted negatives)": (p_tri, True),
    "Symmetric |p - threshold|": (-np.abs(p_tri - best_threshold), False),
    "Random referral": (np.random.default_rng(12345).random(len(y_tri)), False),
}
rows7 = []
for name, (score, only_neg) in policies.items():
    for ref in [0.0, 0.10, 0.20, 0.30]:
        rows7.append({"Method": name, **triage_metrics(score, ref, only_neg)})
triage_df = pd.DataFrame(rows7)
for col, title in [("Silent_miss", "Silent miss rate (share of all diabetics missed)"),
                   ("Auto_error", "Error rate on automatically handled cases"),
                   ("Referred_diabetic_share", "Diabetic share among referred cases")]:
    print("\n" + title)
    print(triage_df.pivot(index="Referral", columns="Method", values=col).round(3).to_string())

# 8. Deep ensemble (5 seeds): epistemic uncertainty + referral figure
import tensorflow as tf
from tensorflow import keras
from scipy.stats import rankdata

ENS_PATH = DRIVE_DIR + "ensemble_probs.npz"
SEEDS = [0, 1, 2, 3, 4]

def build_mlp(n_features, seed):
    tf.keras.utils.set_random_seed(seed)
    m = keras.Sequential([
        keras.layers.Input(shape=(n_features,)),
        keras.layers.Dense(128, activation="relu"), keras.layers.Dropout(0.3),
        keras.layers.Dense(64, activation="relu"), keras.layers.Dropout(0.3),
        keras.layers.Dense(1, activation="sigmoid")])
    m.compile(optimizer=keras.optimizers.Adam(1e-3), loss="binary_crossentropy")
    return m

if os.path.exists(ENS_PATH):
    npz = np.load(ENS_PATH)
    ens_val_members, ens_test_members = npz["val"], npz["test"]
    print("Loaded saved ensemble predictions:", ENS_PATH)
else:
    val_list, test_list = [], []
    for s in SEEDS:
        m = build_mlp(X_train_scaled.shape[1], s)
        m.fit(X_train_scaled, y_train_dl, validation_data=(X_val_scaled, y_val_dl),
              epochs=40, batch_size=512, verbose=0,
              callbacks=[keras.callbacks.EarlyStopping(patience=4, restore_best_weights=True)])
        val_list.append(m.predict(X_val_scaled, batch_size=4096, verbose=0).ravel())
        test_list.append(m.predict(X_test_scaled, batch_size=4096, verbose=0).ravel())
        print(f"seed {s}: test ROC-AUC {roc_auc_score(y_test_dl, test_list[-1]):.4f}")
    ens_val_members, ens_test_members = np.array(val_list), np.array(test_list)
    np.savez(ENS_PATH, val=ens_val_members, test=ens_test_members)

yv_e = np.asarray(y_val_dl).astype(int)
yt_e = np.asarray(y_test_dl).astype(int)
ens_pv, ens_sdv = ens_val_members.mean(0), ens_val_members.std(0)
ens_pt, ens_sdt = ens_test_members.mean(0), ens_test_members.std(0)

aucs = [roc_auc_score(yt_e, m) for m in ens_test_members]
prs = [average_precision_score(yt_e, m) for m in ens_test_members]
print(f"Members ROC-AUC: {np.mean(aucs):.4f} +/- {np.std(aucs):.4f} | "
      f"PR-AUC: {np.mean(prs):.4f} +/- {np.std(prs):.4f}")
print(f"Ensemble ROC-AUC {roc_auc_score(yt_e, ens_pt):.4f} | "
      f"PR-AUC {average_precision_score(yt_e, ens_pt):.4f} | "
      f"Brier {brier_score_loss(yt_e, ens_pt):.5f} | ECE {ece_score(yt_e, ens_pt):.4f}")

ts = np.linspace(0.05, 0.60, 111)
ens_thr = float(ts[int(np.argmax([f1_score(yv_e, (ens_pv >= t).astype(int), zero_division=0)
                                  for t in ts]))])
ens_pred_v = (ens_pv >= ens_thr).astype(int)
ens_pred_t = (ens_pt >= ens_thr).astype(int)
neg_e = ens_pred_t == 0
print(f"Ensemble threshold (validation F1-max): {ens_thr:.3f}")
print("\nSpearman(sd, p):", round(spearmanr(ens_sdt, ens_pt)[0], 4))
print("Missed-diabetic AUROC among predicted negatives | "
      f"p: {roc_auc_score(yt_e[neg_e], ens_pt[neg_e]):.4f} | "
      f"sd: {roc_auc_score(yt_e[neg_e], ens_sdt[neg_e]):.4f}")

def combo(p, s, w):
    return (1 - w) * rankdata(p) / len(p) + w * rankdata(s) / len(s)

def silent_miss_e(y_, pred_, score, ref, only_neg=True):
    pool = np.where(pred_ == 0)[0] if only_neg else np.arange(len(y_))
    k = int(round(ref * len(y_)))
    refer = np.zeros(len(y_), bool)
    if k > 0:
        refer[pool[np.argsort(-score[pool])[:k]]] = True
    return np.sum(~refer & (pred_ == 0) & (y_ == 1)) / y_.sum()

ws = [0, 0.25, 0.5, 0.75, 1.0]
val_scores = {w: silent_miss_e(yv_e, ens_pred_v, combo(ens_pv, ens_sdv, w), 0.20) for w in ws}
w_star = min(val_scores, key=val_scores.get)
print("Validation silent miss @20% referral by weight on sd:",
      {w: round(float(v), 3) for w, v in val_scores.items()}, "-> w* =", w_star)

rows8 = []
for name, w in [("p only (w=0)", 0.0), ("epistemic sd only (w=1)", 1.0),
                (f"combined (w*={w_star})", w_star)]:
    sc = combo(ens_pt, ens_sdt, w)
    rows8.append({"Score": name, **{f"ref {int(r * 100)}%": round(float(silent_miss_e(
        yt_e, ens_pred_t, sc, r)), 3) for r in [0.1, 0.2, 0.3]}})
ensemble_triage_df = pd.DataFrame(rows8)
print("\nTest silent miss rate (no referral:",
      round(float(silent_miss_e(yt_e, ens_pred_t, ens_pt, 0.0)), 3), ")")
print(ensemble_triage_df.to_string(index=False))

refs = np.linspace(0, 0.5, 26)
rand_s = np.random.default_rng(12345).random(len(yt_e))
fig_curves = [
    ("Risk-band (p, predicted negatives)", ens_pt, True, BLUE, "-"),
    ("Symmetric |p - threshold|", -np.abs(ens_pt - ens_thr), False, ORANGE, "--"),
    ("Ensemble epistemic sd", ens_sdt, True, AQUA, "-"),
    ("Random referral", rand_s, False, MUTED, ":"),
]
fig, ax = plt.subplots(figsize=(6.2, 4.6))
for name, score, only_neg, col, ls in fig_curves:
    ys_ = [silent_miss_e(yt_e, ens_pred_t, score, r, only_neg) for r in refs]
    ax.plot(refs * 100, ys_, color=col, ls=ls, lw=2, label=name)
for r in [10, 20, 30]:
    ax.axvline(r, color=GRID, lw=0.8, zorder=0)
ax.set_xlabel("Referred for clinician review (% of all cases)")
ax.set_ylabel("Silent miss rate\n(share of diabetics missed)")
ax.set_title("Referral vs. silent misses (test set)", loc="left", fontsize=12)
ax.set_xlim(0, 50); ax.set_ylim(0, None)
ax.legend(frameon=False, loc="upper center", bbox_to_anchor=(0.5, -0.2), ncol=2)
fig.savefig(FIG_DIR + "referral_vs_silent_miss.png", dpi=300, bbox_inches="tight")
plt.show()

# 9. Class-conditional conformal triage, subgroups, model comparison
A0 = 0.10

pv_c = np.asarray(val_prob_cal).reshape(-1)
pt_c = np.asarray(test_prob_cal).reshape(-1)
yv_c = np.asarray(y_val_dl).astype(int)
yt_c = np.asarray(y_test_dl).astype(int)

def conformal_thresholds(p_val, y_val, a1, a0):
    s1 = np.sort(1 - p_val[y_val == 1]); n1 = len(s1)
    s0 = np.sort(p_val[y_val == 0]);     n0 = len(s0)
    k1 = min(int(np.ceil((n1 + 1) * (1 - a1))), n1)
    k0 = min(int(np.ceil((n0 + 1) * (1 - a0))), n0)
    return s1[k1 - 1], s0[k0 - 1]

def triage_sets(p, q1, q0):
    in1 = (1 - p) <= q1
    in0 = p <= q0
    auto_neg = in0 & ~in1
    auto_pos = in1 & ~in0
    refer = ~(auto_neg | auto_pos)
    return auto_neg, auto_pos, refer

rows9 = []
for a1 in [0.30, 0.20, 0.10, 0.05]:
    q1, q0 = conformal_thresholds(pv_c, yv_c, a1, A0)
    an, ap, rf = triage_sets(pt_c, q1, q0)
    rows9.append({
        "alpha (max miss)": a1,
        "Silent_miss (test)": round(np.sum(an & (yt_c == 1)) / yt_c.sum(), 4),
        "Referral": round(rf.mean(), 3), "Auto_neg": round(an.mean(), 3),
        "Auto_pos": round(ap.mean(), 3),
        "Auto_pos_precision": round(yt_c[ap].mean(), 3) if ap.any() else np.nan})
conformal_df = pd.DataFrame(rows9)
print(conformal_df.to_string(index=False))

alphas = np.arange(0.02, 0.52, 0.02)
miss, refer_rate = [], []
for a1 in alphas:
    q1, q0 = conformal_thresholds(pv_c, yv_c, a1, A0)
    an, ap, rf = triage_sets(pt_c, q1, q0)
    miss.append(np.sum(an & (yt_c == 1)) / yt_c.sum())
    refer_rate.append(rf.mean())
fig, axs = plt.subplots(1, 2, figsize=(10, 4.2))
axs[0].plot([0, 0.5], [0, 0.5], ls=":", color=MUTED, lw=1.5, label="Guarantee (miss <= alpha)")
axs[0].plot(alphas, miss, color=BLUE, lw=2, label="Achieved (test)")
axs[0].set_xlabel("Target alpha (max share of diabetics missed)")
axs[0].set_ylabel("Silent miss rate")
axs[0].set_title("Guarantee holds on test", loc="left", fontsize=12)
axs[0].legend(frameon=False, loc="upper left")
axs[1].plot(alphas, np.array(refer_rate) * 100, color=ORANGE, lw=2)
axs[1].set_xlabel("Target alpha (max share of diabetics missed)")
axs[1].set_ylabel("Referred cases (% of all)")
axs[1].set_title("Workload cost of the guarantee", loc="left", fontsize=12)
fig.tight_layout()
fig.savefig(FIG_DIR + "conformal_guarantee.png", dpi=300, bbox_inches="tight")
plt.show()

def make_groups(Xdf):
    Xdf = Xdf.reset_index(drop=True)
    return {
        "Sex": Xdf["Sex"].map({0: "Female", 1: "Male"}).astype("category"),
        "Age": pd.cut(Xdf["Age"], [0, 5, 9, 13], labels=["18-44", "45-64", "65+"]),
        "Income": pd.cut(Xdf["Income"], [0, 4, 6, 8], labels=["<25k", "25-50k", ">=50k"]),
        "Education": pd.cut(Xdf["Education"], [0, 3, 4, 6], labels=["<HS", "HS grad", "College"])}
grp_te, grp_va = make_groups(X_te_raw), make_groups(X_va_raw)

A1 = 0.10
q1, q0 = conformal_thresholds(pv_c, yv_c, A1, A0)
an, ap, rf = triage_sets(pt_c, q1, q0)
def miss_rate(an_, y_):
    return np.sum(an_ & (y_ == 1)) / max(np.sum(y_ == 1), 1)

grows = []
for dim, s_te in grp_te.items():
    s_va = grp_va[dim]
    for lvl in s_te.cat.categories:
        m_te = (s_te == lvl).values
        m_va = (s_va == lvl).values
        qg1, qg0 = conformal_thresholds(pv_c[m_va], yv_c[m_va], A1, A0)
        anG, apG, rfG = triage_sets(pt_c[m_te], qg1, qg0)
        yg = yt_c[m_te]
        grows.append({
            "Dim": dim, "Group": lvl, "n": int(m_te.sum()), "Prev": round(yg.mean(), 3),
            "AUC": round(roc_auc_score(yg, pt_c[m_te]), 3),
            "ECE": round(ece_score(yg, pt_c[m_te]), 4),
            "Miss_marginal": round(miss_rate(an[m_te], yg), 3),
            "Miss_group_cond": round(miss_rate(anG, yg), 3),
            "Refer_marginal": round(rf[m_te].mean(), 3),
            "Refer_group_cond": round(rfG.mean(), 3)})
subgroup_df = pd.DataFrame(grows)
print(f"\nSubgroup trust (target alpha = {A1}):")
print(subgroup_df.to_string(index=False))

cands = {"LR (unweighted)": (lr_val_p, lr_plain_prob),
         "LightGBM": (gbm_val, gbm_prob),
         "DL Calibrated": (pv_c, pt_c)}
rows9d = []
for name, (pv_m, pt_m) in cands.items():
    for a1 in [0.20, 0.10, 0.05]:
        q1_, q0_ = conformal_thresholds(pv_m, yv_c, a1, A0)
        an_, ap_, rf_ = triage_sets(pt_m, q1_, q0_)
        rows9d.append({"Model": name, "alpha": a1,
                       "Silent_miss": np.sum(an_ & (yt_c == 1)) / yt_c.sum(),
                       "Referral": rf_.mean()})
model_agnostic_df = pd.DataFrame(rows9d)
print("\nReferral rate (lower is better):")
print(model_agnostic_df.pivot(index="alpha", columns="Model", values="Referral").round(3).to_string())
print("\nSilent miss rate on test (should be <= alpha):")
print(model_agnostic_df.pivot(index="alpha", columns="Model", values="Silent_miss").round(3).to_string())

# 10. Guarantee stability (200 splits) + cost-derived triage
rngA = np.random.default_rng(0)
resA = []
for _ in range(200):
    perm = rngA.permutation(len(pt_c))
    cal, ev = perm[: len(pt_c) // 2], perm[len(pt_c) // 2:]
    for a1 in [0.10, 0.05]:
        q1_, q0_ = conformal_thresholds(pt_c[cal], yt_c[cal], a1, A0)
        an_, ap_, rf_ = triage_sets(pt_c[ev], q1_, q0_)
        resA.append({"alpha": a1,
                     "miss": np.sum(an_ & (yt_c[ev] == 1)) / yt_c[ev].sum(),
                     "refer": rf_.mean()})
dfA = pd.DataFrame(resA)
stability_df = dfA.groupby("alpha").agg(
    miss_mean=("miss", "mean"),
    miss_p2_5=("miss", lambda s: np.percentile(s, 2.5)),
    miss_p97_5=("miss", lambda s: np.percentile(s, 97.5)),
    refer_mean=("refer", "mean")).round(4)
print("Guarantee stability (200 random splits):")
print(stability_df.to_string())

def run_policy(p, y, c_fn, c_fp, c_ref, t1, t2):
    an = p < t1
    ap = p > t2
    rf = ~(an | ap)
    realized = (c_fn * np.sum(an & (y == 1)) + c_fp * np.sum(ap & (y == 0))
                + c_ref * rf.sum()) / len(y) * 1000
    predicted = (np.sum(c_fn * p[an]) + np.sum(c_fp * (1 - p[ap]))
                 + c_ref * rf.sum()) / len(y) * 1000
    return realized, predicted, rf.mean()

def rule_thresholds(c_fn, c_fp, c_ref):
    t1, t2 = c_ref / c_fn, 1 - c_ref / c_fp
    if t1 >= t2:
        t1 = t2 = c_fp / (c_fp + c_fn)
    return t1, t2

models10 = {"DL Calibrated": (pv_c, pt_c), "LightGBM": (gbm_val, gbm_prob),
            "LR (unweighted)": (lr_val_p, lr_plain_prob), "LR (balanced)": (lr_bal_val, lr_bal_prob)}
grid = np.linspace(0.01, 0.9, 45)
rowsB = []
for c_fn in [10, 20, 50]:
    c_fp, c_ref = 2, 1
    prev = yt_c.mean()
    trivial = min(c_ref * 1000, c_fn * prev * 1000, c_fp * (1 - prev) * 1000)
    for name, (pv_m, pt_m) in models10.items():
        t1, t2 = rule_thresholds(c_fn, c_fp, c_ref)
        real, pred_, ref_ = run_policy(pt_m, yt_c, c_fn, c_fp, c_ref, t1, t2)
        best = (np.inf, None)
        for a in grid:
            for b in grid[grid >= a]:
                r, _, _ = run_policy(pv_m, yv_c, c_fn, c_fp, c_ref, a, b)
                if r < best[0]:
                    best = (r, (a, b))
        tuned_test, _, _ = run_policy(pt_m, yt_c, c_fn, c_fp, c_ref, *best[1])
        rowsB.append({"c_FN": c_fn, "Model": name, "Cost_rule": round(real, 1),
                      "Cost_predicted": round(pred_, 1), "Gap": round(real - pred_, 1),
                      "Cost_val_tuned": round(tuned_test, 1), "Referral": round(ref_, 3),
                      "Best_trivial": round(trivial, 1)})
cost_df = pd.DataFrame(rowsB)
print("\nCost per 1000 patients (c_FP=2, c_ref=1; illustrative costs):")
print(cost_df.to_string(index=False))

# 11. Save every result table as CSV
tables = {
    "model_comparison": comparison_df, "ablation": ablation_df, "gbm_comparison": gbm_table,
    "triage_baselines": triage_df, "ensemble_triage": ensemble_triage_df,
    "conformal_guarantee": conformal_df, "subgroup_trust": subgroup_df,
    "model_agnostic": model_agnostic_df, "guarantee_stability": stability_df.reset_index(),
    "cost_analysis": cost_df,
}
for name, df in tables.items():
    df.to_csv(RES_DIR + name + ".csv", index=False)
print("Saved", len(tables), "tables to", RES_DIR)
print("Figures are in", FIG_DIR)
```

```python
# 12. Confidence intervals for the key triage claims + net-benefit (decision-curve) analysis
B12 = 500
rng12 = np.random.default_rng(2024)
y12 = np.asarray(y_test_dl).astype(int)
p12 = np.asarray(test_prob_cal).reshape(-1)
n12 = len(y12)

# 12a. Bootstrap CIs: silent miss rate at fixed referral budgets (test-set resampling)
def silent_miss_policy(p, y, tau, ref, kind, rng):
    n = len(y)
    pred = (p >= tau).astype(int)
    if kind == "risk_band":
        pool = np.where(pred == 0)[0]; score = p
    elif kind == "symmetric":
        pool = np.arange(n); score = -np.abs(p - tau)
    else:
        pool = np.arange(n); score = rng.random(n)
    k = min(int(round(ref * n)), len(pool))
    refer = np.zeros(n, bool)
    if k > 0:
        refer[pool[np.argpartition(-score[pool], k - 1)[:k]]] = True
    return np.sum(~refer & (pred == 0) & (y == 1)) / max(y.sum(), 1)

kinds = {"risk_band": "Risk-band (p, predicted negatives)",
         "symmetric": "Symmetric |p - threshold|", "random": "Random referral"}
rows12, rows12d = [], []
for ref in [0.10, 0.20, 0.30]:
    est = {k_: [] for k_ in kinds}
    for _ in range(B12):
        idx = rng12.integers(0, n12, n12)
        for k_ in kinds:
            est[k_].append(silent_miss_policy(p12[idx], y12[idx], best_threshold, ref, k_, rng12))
    for k_, name in kinds.items():
        lo, hi_ = np.percentile(est[k_], [2.5, 97.5])
        point = silent_miss_policy(p12, y12, best_threshold, ref, k_, np.random.default_rng(12345))
        rows12.append({"Referral": ref, "Policy": name,
                       "Silent_miss": f"{point:.3f} [{lo:.3f}, {hi_:.3f}]"})
    for other in ["symmetric", "random"]:
        d = np.array(est["risk_band"]) - np.array(est[other])
        lo, hi_ = np.percentile(d, [2.5, 97.5])
        rows12d.append({"Referral": ref, "Comparison": f"Risk-band minus {kinds[other]}",
                        "Diff": f"{d.mean():+.3f} [{lo:+.3f}, {hi_:+.3f}]",
                        "Share_of_resamples_risk_band_better": round(float((d < 0).mean()), 3)})
triage_ci_df = pd.DataFrame(rows12)
triage_diff_df = pd.DataFrame(rows12d)
print("Silent miss rate with bootstrap 95% CI (test set):")
print(triage_ci_df.to_string(index=False))
print("\nPaired difference (negative = risk-band misses fewer diabetics):")
print(triage_diff_df.to_string(index=False))

# 12b. Subgroup bootstrap CIs for the silent miss rate (test sampling uncertainty only;
#      thresholds are fixed from the validation set)
q1m, q0m = conformal_thresholds(pv_c, yv_c, A1, A0)
an_m, _, _ = triage_sets(pt_c, q1m, q0m)
rows12b = []
for dim, s_te in grp_te.items():
    s_va = grp_va[dim]
    for lvl in s_te.cat.categories:
        m_te = (s_te == lvl).values
        m_va = (s_va == lvl).values
        qg1, qg0 = conformal_thresholds(pv_c[m_va], yv_c[m_va], A1, A0)
        an_g, _, _ = triage_sets(pt_c[m_te], qg1, qg0)
        yg, am = yt_c[m_te], an_m[m_te]
        ng = len(yg)
        bm, bg = [], []
        for _ in range(B12):
            ii = rng12.integers(0, ng, ng)
            bm.append(miss_rate(am[ii], yg[ii])); bg.append(miss_rate(an_g[ii], yg[ii]))
        lo_m, hi_m = np.percentile(bm, [2.5, 97.5])
        lo_g, hi_g = np.percentile(bg, [2.5, 97.5])
        rows12b.append({
            "Dim": dim, "Group": lvl, "n": ng, "Diabetics": int(yg.sum()),
            "Miss_marginal": f"{miss_rate(am, yg):.3f} [{lo_m:.3f}, {hi_m:.3f}]",
            "Miss_group_cond": f"{miss_rate(an_g, yg):.3f} [{lo_g:.3f}, {hi_g:.3f}]",
            "Marginal_CI_above_alpha": bool(lo_m > A1),
            "Marginal_CI_below_alpha": bool(hi_m < A1)})
subgroup_ci_df = pd.DataFrame(rows12b)
print(f"\nSubgroup silent miss rate with 95% CI (target alpha = {A1}):")
print(subgroup_ci_df.to_string(index=False))

# 12c. Net benefit (decision-curve analysis) for the calibrated models
def net_benefit(y, p, pt):
    pred = p >= pt
    tp = np.sum(pred & (y == 1)); fp = np.sum(pred & (y == 0))
    return tp / len(y) - fp / len(y) * pt / (1 - pt)

pts = np.linspace(0.02, 0.50, 49)
prev12 = y12.mean()
nb_models = {"DL Calibrated": p12, "LightGBM": gbm_prob, "LR (unweighted)": lr_plain_prob}
nb_df = pd.DataFrame({"threshold": pts,
                      **{n_: [net_benefit(y12, p_, t) for t in pts] for n_, p_ in nb_models.items()}})
nb_df["Treat all"] = prev12 - (1 - prev12) * pts / (1 - pts)
nb_df["Treat none"] = 0.0
show = nb_df[np.isclose(nb_df["threshold"].values[:, None],
                        [0.05, 0.10, 0.15, 0.20, 0.25, 0.30]).any(axis=1)].round(4)
print("\nNet benefit (per patient) at selected risk thresholds:")
print(show.to_string(index=False))

fig, ax = plt.subplots(figsize=(6.2, 4.6))
for name, col, ls in [("DL Calibrated", BLUE, "-"), ("LightGBM", ORANGE, "--"),
                      ("LR (unweighted)", AQUA, "-"), ("Treat all", MUTED, ":")]:
    ax.plot(nb_df["threshold"] * 100, nb_df[name], color=col, ls=ls, lw=2, label=name)
ax.axhline(0, color=MUTED, lw=1, label="Treat none")
ax.set_ylim(-0.02, prev12 * 1.1)
ax.set_xlim(2, 50)
ax.set_xlabel("Risk threshold (%)")
ax.set_ylabel("Net benefit per patient")
ax.set_title("Decision curve (test set)", loc="left", fontsize=12)
ax.legend(frameon=False, loc="upper center", bbox_to_anchor=(0.5, -0.2), ncol=3)
fig.savefig(FIG_DIR + "decision_curve.png", dpi=300, bbox_inches="tight")
plt.show()

# 12d. Save the new tables
for name, df in {"triage_ci": triage_ci_df, "triage_diff": triage_diff_df,
                 "subgroup_ci": subgroup_ci_df, "net_benefit": nb_df}.items():
    df.to_csv(RES_DIR + name + ".csv", index=False)
print("\nSaved 4 more tables to", RES_DIR)
```
